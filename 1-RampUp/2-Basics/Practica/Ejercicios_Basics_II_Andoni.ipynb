## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

$$radianes = grados*(pi/180)$$

````python
import math

grados = int(input("Introducir numero de grados: "))
print("Los radiantes son: ", grados * (math.pi / 180))
````

## Ejercicio 5
Crea un programa donde se recojan dos inputs del usuario, y el output del programa sea si esos inputs son iguales o no

````python
salida = True
while salida:
    n1 = input("introducir un primer numero: ")
    n2 = input("Introducir un segundo numero")
    if n1 == n2:
        print("Ambos numeros introducidos son iguales")
    else: 
        print("Los números no son iguales")
    salir = int(input("introducir 1 para finalizar el programa."))
    if salir == 1:
        salida = False
print("Fin del programa")
        
````

## Ejercicio 6
Mismo programa que en 5, pero en esta ocasión tienen que ser tres inputs y dos salidas. Una de las salidas que nos indique si todos son iguales, y la otra si al menos dos inputs sí que lo son

````python
numero_1 = input("Introducir primer numero")
numero_2 = input("Introducir segundo numero")
numero_3 = input("introducir tercer numero")

if numero_1 == numero_2 and numero_1 == numero_3:
    print("todos los numeros son iguales")
elif numero_1 == numero_2 or numero_1 == numero_3 or numero_2 == numero_3:
    print("Al menos dos de los numeros son iguales")
else:
    print("hay algun número que no es igual al resto")
````

## Ejercicio 7
Crea un programa que recoja dos inputs. Tiene que comprobar si su suma es igual, superior o inferior a 10, cada condición por separado

````python
numero_1 = float(input("Introducir un número"))
numero_2 = float(input("Introducir otro número"))
if numero_1 + numero_2 == 10:
    print("suma igual a 10")
elif numero_1 + numero_2 > 10:
    print("suma superior a 10")
else:
    print("suma inferior a 10")
````

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "Lo primero de todo, cómo están los máquinas"

1. Pasarlo todo a mayúsculas
2. Pasarlo todo a minúsculas
3. Pasar a mayúsculas solo las iniciales de las palabras
4. Crea una lista dividiendolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

````python
texto = "Lo primero de todo, cómo están los máquinas"

mayus = texto.upper()
minus = texto.lower()
primera_letra_mayus = texto.title()
separado = texto.split(" ")
sustituido = texto.replace(",", ";")
texto_sin_a = texto.replace("a", "")
        
        
print(mayus, minus, primera_letra_mayus, separado, sustituido, texto_sin_a)
````
