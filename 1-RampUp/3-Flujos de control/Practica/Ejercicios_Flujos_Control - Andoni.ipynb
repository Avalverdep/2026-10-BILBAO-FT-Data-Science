## Ejercicio 1
Declara una variable numérica que será una hora cualquiera del día. Implementa la siguiente casuística:

* Si es entre las 0 y las 8, print "Durmiendo"
* Si es entre las 9 y las 18, print "Trabajando"
* Si es entre las 19 y las 21, print "Clase"
* Si es entre las 22 y las 24, print "Descanso"
* En cualquier otro caso, print "Transporte o error"

```python
hora = 0
elegido = True
while elegido:
    hora = int(input("introducir una hora entre las 0 y las 24"))
    if hora < 0 or hora > 24:
        print("introducir una hora correcta")
        hora = 0
    else:
        elegido = False


if hora >= 0 and hora <= 8:
    print("durmiendo")
elif hora >= 9 and hora <= 18:
    print("trabajando")
elif hora >= 19 and hora <= 21:
    print("clase")
elif hora >= 22 and hora <= 24:
    print("descanso")

```

## Ejercicio 2
En este ejercicio vamos a implementar un calculador de precios de casas muy sencillo. Tenemos las siguientes variables:

> superficie
>
> distrito

Implementa la siguiente casuística:

1. Si el distrito es "Moncloa" o "Centro", y además la superficie es superior a 100 metros cuadrados, el precio de la casa es de 1000
2. Si el distrito es "Salamanca", y además la superficie de la casa es al menos de 150 metros, el precio de la casa es de 1500
3. Si el distrito no es "Retiro" y la superficie está entre 60 y 80 metros, el precio es de 600
4. En cualquier otro caso, el precio será de 0

```python

superficie = float(input("Introduce la cantidad de metros de superficie "))
distrito = input("Introduce el nombre del distrito donde consultar: ").lower().strip()

match distrito:
    case  "moncloa" | "centro" if superficie > 100: print("Precio de la casa de 1000€")
    case  "salamanca" if superficie >= 150: print("Precio 1500 €")
    case  "retiro" if superficie > 60 and superficie < 80: print("Precio de 600€")
    case _: print("Ningún resultado compatible con la búsqueda")
```

## Ejercicio 6
Dada la siguiente lista:

> list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

Implementa un programa que los recorra e imprima por pantalla todos los divisibles por 5. Si nos encontramos con alguno que sea mayor que 150, detener el bucle.

```python
list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

for numero in list1:
    if numero % 5 == 0:
        print(numero)
        if numero > 150:
            break
```

## Ejercicio 9
En este ejercicio vamos a crear un pequeño juego. Se trata de intentar adivinar un numero del 1 al 5. Tenemos dos intentos para acertar. Pasos a seguir:

1. Ya viene implementado con la librería `random` cómo obtener un número aleatorio del 1 al 5
2. Tendrás que declarar en una variable el numero de vidas, y mediante un bucle while, comprobar que todavia quedan vidas.
3. Dentro del bucle, obtener el valor del usuario y comprobar si es ese el numero a adivinar. Si no, actualizar las vidas.
4. Si acertamos, salimos del bucle e imprimimos por pantalla "You win". Y si perdemos también salimos del bucle, pero en este caso imprimimos por pantalla "You lose".

TIP: te puede resultar útil usar la sentencia `else` cuando acabe el bucle `while`. Lo que haya dentro de ese `else` se ejecutará una vez acabe la ejecución del `while`. Lo podrás usar para cuando pierdas.

```python
import random as rnd
vida = 5
numero = rnd.randint(1, 5)
print(numero)
while True:
    if vida == 0:
        print("Te has quedado sin vida")
        break
    respuesta = int(input("Intenta adivinar el número: "))
    if respuesta == numero:
        print("VICTORIA")
        break
    else:
        print("respuesta incorrecta, sigue intentandolo")
        vida -= 1
       
        

```
