## Ejercicio Extra
Programa una calculadora que

* Pregunte qué operación quieres realizar (multiplicación/división/suma/resta/potencia/salir)
* Pida dos números 
* Realice la operación
* Siga funcionando hasta que el usuario elija salir

#### Bonus:
* Que guarde todas las operaciones realizadas y las imprima por pantalla cuando el usuario elija esa opción
* Que sea robusta frente a entradas erróneas del usuario o frente a operaciones como dividir por 0

```python
import math
import pprint
import time

salida = True
historico = []
contador = 1
while salida:
    resultado = 0
    entrada_us = int(input("Elegir qué operación realizar: 1 - suma, 2 - resta, 3 - potencia, 4 - división, 5 - salir, 6 - mostrar histórico"))
    match entrada_us:
        case 1:
            n1 = float(input("Introducir número 1: "))
            n2 = float(input("introducir número 2: "))
            resultado = n1 + n2
            guardar_historico = "Operacion numero " , str(contador), "Suma", resultado
            historico.append(guardar_historico)
            contador += 1
            print("Resultado de la suma: ", resultado)
        case 2:
            n1 = float(input("Introducir número 1: "))
            n2 = float(input("introducir número 2: "))
            resultado = n1 - n2
            guardar_historico = "Operacion numero: ", str(contador), "Resta: " + str(resultado)
            historico.append(guardar_historico)
            contador += 1
            print("resultado de la resta: ", resultado)
        case 3: 
            n1 = float(input("introducir un número"))
            n2 = float(input("introducir un valor para elevar:"))
            resultado = math.pow(n1, n2)
            guardar_historico = "Operacion numero: ", str(contador), "Potencia: " + str(resultado)
            historico.append(guardar_historico)
            contador +=1
            print("el resultado de la operación es: ", resultado)
        case 4:
            n1 = float(input("introducir el primer número"))
            n2 = float(input("introducir el segundo número"))
            try:
                resultado = n1 / n2
                guardar_historico =  "Operacion numero: ", str(contador),"Division: " + str(resultado)
                historico.append(guardar_historico)
                contador +=1
                print(resultado)
            except Exception:
                 print("Error en la operación")
        case 5:
            print("Apaganado calculadora")
            salida = False

        case 6:
            print(historico)

        
```
