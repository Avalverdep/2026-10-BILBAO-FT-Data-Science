## Ejercicio 1
Somos una empresa que tiene varias tiendas de electrodomésticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:
* Nombre clase: "Tienda"
* Atributos comunes:
    * Tipo: "Electrodomésticos"
    * Abierta: True
* Atributos propios de cada tienda:
    * Nombre - String
    * Dirección - String
    * Número de empleados - int
    * Ventas últimos 3 meses - Lista de 3 numeros
* Método para calcular las ventas de todos los meses, que devuelva un número con todas las ventas.
* Método que calcula la media de ventas de los últimos meses, por empleado
* Método que devuelve en un string el nombre de la tienda, junto con su dirección.
* Método que obtiene las ventas del último mes.
* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

Se pide:
1. Implementar la clase "Tienda"
2. Crear tres tiendas con datos inventados
3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)
3. Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle `for`
4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
5. Seguro que ya has hecho este apartado... Documenta la clase :)

```python
class Tienda:
    Tipo:str = "Electrodomésticos"
    Abierta:bool = True

    '''Clase tienda, tiene como atributos predefinidos "Tipo", y el buleano "Abierta"'''

    def __init__(self, Nombre:str, Direccion:str, Num_empleados:int, Ventas:list):
        '''Constructor, requiere establecer nombre, dirección y el núnmero total de empleados, requiere por último una list con el total de ventas mensual'''
        if not isinstance(Nombre, str):
            raise TypeError("Nombre debe ser un string")
        if not isinstance(Direccion, str):
            raise TypeError("Direccion debe ser un string")
        if not isinstance(Num_empleados, int):
            raise TypeError("Num_empleados debe ser un int")
        if not isinstance(Ventas, list):
            raise TypeError("Las ventas debe ser una lista")
        self.Nombre = Nombre
        self.Direccion = Direccion
        self.Num_empleados = Num_empleados
        self.Ventas = Ventas

    def CalcularVentas(self):
        '''Método para calcular el total de ventas de la lista que se ha pasado por argumento a la hora de crear el objeto'''
        return sum(self.Ventas)

    def posicion(self):
        '''Este método devuelve un string con el nombre del objeto y la dirección de la tienda'''
        return f"{self.Nombre} en la dirección: {self.Direccion}"

    def CalcularMedia(self):
        '''Cálculo de la media usando la suma de las ventas entre los meses (nunmero de elementos en la lista) y entre la  cantidad total de empleados establecida'''
        resultado = sum(self.Ventas) / len(self.Ventas) / self.Num_empleados
        return resultado

    def Ventas_ultimo_mes(self):
        '''Método para calcular el total de las ventas del último mes (último elemento dentro de la lista de ventas asignada)'''
        resultado = 0
        for i, venta in enumerate(self.Ventas):
            if i == len(self.Ventas) - 1:
                resultado = venta
        return resultado

    def ProyeccionVentas(self, inversion):
        '''método para calcular la proyección de ventas en relación a la variable que se introduzca'''
        if inversion < 1000:
           return  sum(self.Ventas) * 1.2
        elif inversion >= 1000:
            return sum(self.Ventas) * 1.5
        
        
ventas_1 = [1250, 3200, 680]
tienda_1 = Tienda("Tienda 1", "Calle Falsa, 1,2,3", 5, ventas_1)
ventas_2 = [5000, 3380, 5200]
tienda_2 = Tienda("Tienda 2", "Avenida Falsa, 3,2,1", 8, ventas_2)
ventas_3 = [550, 1200, 189]
tienda_3 = Tienda("Tienda 3", "Calle Verdadera, 8", 3, ventas_3)
name = tienda_2.Nombre
print(f"Los datos de {tienda_2.posicion()}, la media de sus ventas es: {tienda_2.CalcularMedia()}, el total de las ventas es: {tienda_2.CalcularVentas()}, las ventas del ultimo mes son: {tienda_2.Ventas_ultimo_mes()}, y la proyeccion de ventas, si la proyeccion fuesen 6000 € es: {tienda_2.ProyeccionVentas(6000)}")


def CalcularTotalVentas (t1 = Tienda, t2 = Tienda, t3 = Tienda):
    total = 0
    for tienda in t1 , t2, t3:
        total += tienda.CalcularMedia()
    return total


print(f"El total de las medias de las tres tiendas es: {CalcularTotalVentas(tienda_1, tienda_2, tienda_3)}")

def imprimir_nombres(t1 = Tienda, t2 = Tienda, t3 = Tienda):
    resultado = ""
    for tienda in t1, t2, t3:
        if tienda.posicion().__contains__("Avenida"):
            resultado += f"En la dirección de {tienda.Nombre} hay la palabra 'Avenida': {tienda.posicion()}"
    if resultado:
        return resultado
    else:
        return "No hay ninguna coincidencia"
    

print(imprimir_nombres(tienda_1, tienda_2, tienda_3))
```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
class Perro:
    ''' Clase perro con atributos por defecto: orejas, ojos y velocidad.'''
    orejas = 4
    ojos = 2
    velocidad = 0

    def __init__(self, raza, pelo="Marrón", duenio=False): 
        '''Constructor de la clase Perro, recibe raza, pelo y duenio como boleano, por defecto tiene el pelo marrón, y sin duenio'''
        self.raza = raza
        self.pelo = pelo
        self.duenio = duenio

    def andar(self, aumento_velocidad): 
        '''Aumenta la velocidad (por defecto 0)'''
        self.velocidad += aumento_velocidad

    def dejar_de_andar(self): 
        '''restablece la velocidad desde el nivel anterior a 0'''
        self.velocidad = 0

    def ladrar(self, ladrido:str):
        '''devuelve un string con el termino que se introduce  por arumento''' 
        string = "GUAU! " , ladrido
        return string

perrillo = Perro("Mastín", "Negro")
print(perrillo.raza, perrillo.pelo, perrillo.duenio)
perrillo.andar(500)
print(perrillo.velocidad)
perrillo.dejar_de_andar()
print(perrillo.velocidad)
perrillo.andar(500)
print(perrillo.velocidad)
perrillo.dejar_de_andar()
print(perrillo.velocidad)
print(perrillo.ladrar("AAAAAAAAAAAAAARRRRRRRRRRRRRRRRGGGGGGGGGGGGGGGGGGGGHHHHHHHHHHHHHHHHH"))
print(perrillo.dejar_de_andar.__doc__)
```
