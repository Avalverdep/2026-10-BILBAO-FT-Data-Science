## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
palabra = input("Introducir una palabra para jugar").strip().upper()


def cambiar_letras(entrada):
    salida = ""
    for i in entrada:
        salida += "_"
    return salida

campo_juego = cambiar_letras(palabra)

def juego(campo, respuesta):
    lista_palabra = []
    lista_campo = []
    for i in respuesta:
        lista_palabra.append(i)
    for j in campo:
        lista_campo.append(j)
    letras_intentadas = []
    intentos = 6 
    while intentos > 0: 
        entrada = ""
        while True:
            entrada = input("Introducir una letra para adivinar la palabra").strip().upper()
            if len(entrada) != 1:
                print("Sólo se pueden introducir letras, no palabras enteras.")
            elif entrada not in letras_intentadas:
                letras_intentadas.append(entrada)
                break
            else:
                print("Ya has dicho esa letra, intenta otra.")
        if entrada in lista_palabra:
            print("encontrado")
            for x, z in enumerate(lista_palabra):
                if z == entrada:
                    lista_campo[x] = entrada  
            mostrar = ""
            for i in lista_campo:
                mostrar += i +" "
            print(mostrar)
            if lista_campo == lista_palabra:
                print("Has ganado")
                intentos = 0
                break  
        else:
            if intentos == 2:
                intentos -= 1
                print(f"Incorrecto, te queda {intentos} intento")
            elif intentos == 0:
                print("Has perdido, ya no te quedan intentos para seguir.")
            else:
                intentos -= 1
                print(f"Incorrecto, te quedan {intentos} intentos")
                
            

campo_juego = cambiar_letras(palabra)
juego(campo_juego, palabra)


        
        

````
