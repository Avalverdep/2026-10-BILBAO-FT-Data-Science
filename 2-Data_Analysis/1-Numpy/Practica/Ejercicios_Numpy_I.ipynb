## Ejercicio 5
En un nuevo array, sustituye todos los impares de `my_array` por -1

````python
n_array = my_array.copy()
n_array[n_array % 2 != 0] = -1
print(n_array)
````

## Ejercicio 6
En un nuevo array, sustituye todos los impares de `my_array` por -1. Esta vez usa `where`

````python
n_array_2 = my_array.copy()
n_array_2 = np.where(n_array_2 % 2 != 0, -1, n_array_2)
print(n_array_2)
````
