## Ejercicio 8
Dado el siguiente array:

```Python
x = np.array(["Loro", "Perro", "Gato", "Loro", "Perro"])
```

Filtra el array para quedarte únicamente con los loros.

Resuélvelo de dos maneras diferentes

````python
array = np.array(["Loro", "Perro", "Gato", "Loro", "Perro"])
print(np.array(list(filter(lambda x : x == "Loro", array))))
arr_filtrado = array[np.where(array == "Loro")]
print(arr_filtrado)
````

## Ejercicio 12
Encuentra los índices de los valores que no sean 0 del siguiente vector [1,2,0,0,4,0]. El output sería [1, 2 4]. [Encuentra el método apropiado en la documentación.](https://numpy.org/doc/) Encapsúlalo en una función, con un argumento de entrada de un vector.

````python
vector = np.array([1, 2, 0, 0, 4, 0])

def encontrar_indices(entrada):
    sin_ceros = entrada.nonzero()

    return sin_ceros
print(encontrar_indices(vector))
````
