## Ejercicio descargas

Python es un lenguaje de propósito general, por lo que podemos desarrollar otros programas,
aparte de ciencia de datos, que nos permitan montar dashboards, crear APIs o automatizar
procesos. En este caso vas a automatizar un proceso, en el que tendrás que desarrollar un
script que ordene todos los archivos de una carpeta, dependiendo de la extensión de los
mismos.


El objetivo es ordenar la típica carpeta de cajón desastre donde se nos acumulan archivos de
todos los formatos como: imágenes, words, presentaciones, excels, ejecutables, zips, etc... Y
que no vamos a ir viendo archivo a archivo, ordenándolos.... La solución suele ser eliminarlo
todo. Vamos a intentar no tener que llegar a eso, ordenando los ficheros automáticamente en
carpetas y revisar las que nos interesen. La típica carpeta donde aplicaríamos este programa es
la de “Descargas”, donde se suelen acumular muchos archivos temporales.


Por tanto, **el programa tiene que guardar los archivos de la carpeta “Descargas” (o cualquier
otra donde acumules mucho archivo temporal) en los siguientes directorios dentro de la
propia carpeta “Descargas”**:
- Imagenes
- Documentos
- Software
- Otros

Cada vez que ejecutes el script, automáticamente se ordenarán todos los archivos en sus
correspondientes carpetas.


```python
doc_types = ('.doc', '.docx', '.txt', '.pdf', '.xls', '.ppt', '.xlsx', '.pptx')
img_types = ('.jpg', '.jpeg', '.png', '.svg', '.gif', 'webp')
software_types = ('.exe', '.py','.ipynb')
```

### 1. Crea un un programa con el flujo principal y encapsúlalo en funciones.

```python
import os
import shutil

```

```python

os.chdir(r'C:\Users\andon\Downloads')  #Cambiar al nuevo directorio creado para mover los archivos de descargas
os.getcwd() #Comprobar que haya funcionado

```

```python
directorio_n = r'C:\Users\andon\Desktop\Descargas_n'

def MoverImagenes ():
    dir_imagenes = os.path.join(directorio_n, 'img')
    os.makedirs(dir_imagenes, exist_ok=True)
    for i in os.listdir():
        if i.endswith(img_types):
            print("Imagen", i)
            shutil.move(i, dir_imagenes)

def MoverDocs ():
    dir_documentos = os.path.join(directorio_n, 'Documentos')
    os.makedirs(dir_documentos, exist_ok=True)
    for i in os.listdir():
        if i.endswith(doc_types):
            print('Documento: ', i)
            shutil.move(i, dir_documentos)
def MoverSw ():
    dir_sw = os.path.join(directorio_n, 'Software')
    os.makedirs(dir_sw, exist_ok=True)
    for i in os.listdir():
        if i.endswith(software_types):
            print('SW', i)
            shutil.move(i, dir_sw)

def MoverOtros():
    dir_otros = os.path.join(directorio_n, 'Otros')
    os.makedirs(dir_otros, exist_ok=True)
    for i in os.listdir():
        if not i.endswith(img_types) and not i.endswith(software_types) and not i.endswith(doc_types):
            print('Movidas variadas: ', i)
            shutil.move(i, dir_otros)

def MoverTodasLasCosas():   
    MoverImagenes()
    MoverDocs()
    MoverSw()
    MoverOtros()

def DevolverTodoALaNormalidad():
    descargas = os.getcwd()
    for carpeta in ['img', 'Documentos', 'Software', 'Otros']:
        dir_carpeta = os.path.join(directorio_n, carpeta)
        for i in os.listdir(dir_carpeta):
            print('Devolviendo: ', i)
            shutil.move(os.path.join(dir_carpeta, i), descargas)



MoverTodasLasCosas()
DevolverTodoALaNormalidad()


```
